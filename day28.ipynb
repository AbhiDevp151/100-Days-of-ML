```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv("Titanic.csv")
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:4],df.iloc[:, -1],test_size=0.2,random_state=0)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Separate transformer steps directly inside ColumnTransformer:
# ('step_name', TransformerObject, [column_indices_or_names])

transformer = ColumnTransformer(
    transformers=[
        # 1. Impute missing values for numerical (e.g., Age)
        ('impute_age', SimpleImputer(strategy='median'), ['Age']),
        
        # 2. Scale Fare
        ('scale_fare', StandardScaler(), ['Fare']),
        
        # 3. Ordinal encode Pclass
        ('ordinal_pclass', OrdinalEncoder(categories=[[3, 2, 1]]), ['Pclass']),
        
        # 4. One-Hot encode nominal features (Sex, Embarked)
        ('ohe_nominal', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), ['Sex', 'Embarked'])
    ],
    remainder='passthrough'  # Keeps any remaining columns unmodified
)

# Fit on training data and transform both
x_train_transformed = transformer.fit_transform(x_train)
x_test_transformed = transformer.transform(x_test)

print(x_train_transformed.shape)
```
