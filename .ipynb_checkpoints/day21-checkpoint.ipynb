```python
import pandas as pd
import seaborn as sns
```

```python
tips=sns.load_dataset("tips")
#flights=sns.load_dataset("flight")
tips
```

```python
titanic=pd.read_csv("Titanic.csv")
```

# Scattertplot #

```python
sns.scatterplot(x=tips["total_bill"],y=tips["tip"],hue=tips["sex"],style=tips["smoker"])
```

# BARplot #

```python
sns.barplot(data=titanic,x='Pclass',y='Fare',hue='Sex')
```

#  Distplot #

```python
sns.kdeplot(data=titanic, x='Age', hue='Survived', common_norm=False)
```

# heatmap#

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived'], normalize='index') * 100, annot=True, fmt='.1f', cmap='YlGnBu')
```
