# Normalization #

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Wine_data.csv", usecols=[0, 1, 2])
df.columns = ['class label', 'Alcohol', 'Malic acid']
```

```python
df
```

```python
sns.kdeplot(df['Alcohol'])
```

```python
sns.kdeplot(df['Malic acid'])
```

```python
color_dict={1:'red',2:'blue',3:'green'}
sns.scatterplot(x='Alcohol',y='Malic acid',hue='class label',data=df, palette=color_dict)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(df.drop('class label',axis=1),df['class label'],test_size=0.3,random_state=0)
x_train.shape,x_test.shape
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
scaler.fit(x_train)
x_train.scaled=scaler.transform(x_train)
x_test.scaled=scaler.transform(x_test)
```

```python
x_train.scaled=pd.DataFrame(x_train.scaled,columns=x_train.columns)
x_test.scaled=pd.DataFrame(x_test.scaled,columns=x_test.columns)
```

```python
np.round(x_train.describe(),1)
```
