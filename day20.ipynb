## EDA  Univariate Analysis ##

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df=pd.read_csv("customer_churn_practice.csv")
df.sample(5)
```

```python
#sns.countplot(data=df, x='gender', hue='gender', palette='Set2', legend=False)
#df['gender'].value_counts().plot(kind='bar')
sns.countplot(
    data=df,
    x='payment_method',
    hue='payment_method',
    palette='Set2',
    legend=False
)
plt.show()
```

```python
df['payment_method'].value_counts().plot(kind='pie',autopct='%.2f')
```

# Histogram #

```python
sns.distplot(df["age"])
```
