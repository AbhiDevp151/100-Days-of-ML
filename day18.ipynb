# Data Scraping #

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36"
}
webpage=requests.get("https://www.ambitionbox.com/list-of-companies?page=2",headers=headers).text
```

```python
soup = BeautifulSoup(webpage, 'html.parser')
```

```python
#soup.find_all('h2')[2].text.strip()
```

```python
company=soup.find_all('div',class_='companyCardWrapper')
```

```python
def get_text(element):
    try:
        return element.text.strip()
    except AttributeError:
        return None

name, rating, reviews, domain_location = [], [], [], []

for i in company:
    name.append(get_text(i.find('h2')))
    
    # Rating is stored inside div with class 'rating_text'
    rating.append(get_text(i.find('div', class_='rating_text rating_text--md ')))
    
    # Reviews count is stored inside span with class 'companyCardWrapper__companyRatingCount'
    reviews.append(get_text(i.find('span', class_='companyCardWrapper__companyRatingCount')))
    
```

```python
name
```

```python
data = {
    'Company Name': name,
    'Rating': rating
}
df = pd.DataFrame(data)
df.to_csv('ambitionbox_companies.csv', index=False)
df.head()
```
